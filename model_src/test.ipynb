```python
# ============================================
# 🍅 POMIDOR FITAFTOROZ KASALLIGI ANIQLOVCHI
# ============================================

import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image
from google.colab import files
from IPython.display import display, Image as IPImage
from PIL import Image
import io

# --- 1. MODELNI YUKLASH ---
model = load_model('/content/tomato_model.h5')
print('✅ Model yuklandi!')
print(f'   Kirish o\'lchami: {model.input_shape}')
print(f'   Chiqish soni: {model.output_shape}')

# --- 2. CLASS NOMLARI ---
# Alfavit tartibida (ImageDataGenerator shu tartibda o'qiydi)
CLASS_NAMES = [
    '0%   — Sog\'lom, kasallik yo\'q',
    '0.1%-10% — Minimal zarar',
    '11%-25%  — Past darajada zarar',
    '26%-50%  — O\'rtacha zarar',
    '51%+     — Yuqori darajada zarar'
]

# --- 3. BASHORAT FUNKSIYASI ---
def predict(img_path):
    input_shape = model.input_shape
    h = input_shape[1] if input_shape[1] else 224
    w = input_shape[2] if input_shape[2] else 224

    img = Image.open(img_path).convert('RGB').resize((w, h))
    arr = np.array(img) / 255.0
    arr = np.expand_dims(arr, axis=0)

    probs = model.predict(arr, verbose=0)[0]
    idx = np.argmax(probs)

    print('\n' + '━'*45)
    print('📊  NATIJA')
    print('━'*45)
    display(IPImage(img_path, width=280))
    print(f'\n🏷️  Kasallik darajasi : {CLASS_NAMES[idx]}')
    print(f'📈  Ishonch           : {probs[idx]*100:.1f}%')
    print('\n📋  Barcha classlar:')
    for i, (name, p) in enumerate(zip(CLASS_NAMES, probs)):
        bar = '█' * int(p * 25)
        tag = ' ◀ ASOSIY' if i == idx else ''
        print(f'   {p*100:5.1f}%  {bar:<25} {name}{tag}')
    print('━'*45)

# --- 4. RASM YUKLASH VA TEST ---
print('\n📸 Rasm yuklang:')
uploaded = files.upload()
for fname in uploaded:
    predict(fname)
```
